# 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Import the Dataset

```python
df = pd.read_csv("/content/Car Market Trends Analysis with Car Dekho Data.csv")
```

```python
#Step 3: View First 5 Rows

df.head()
```

```python
# Step 4: View Last 5 Rows
df.tail()
```

```python
# Step 5: Check Number of Rows and Columns
df.shape
```

```python
# Step 6: Check Column Names
df.columns
```

```python
# Step 7: Check Data Information
df.info()
```

```python
# Step 8: Check Missing Values
df.isnull().sum()
```

```python
# Step 9: Check Duplicate Rows
df.duplicated().sum()
```

```python
# Step 10: Remove Duplicate Rows
df = df.drop_duplicates()
print("Duplicates removed successfully!")
print("New shape:", df.shape)
```

```python
# Step 11: Basic Statistics
df.describe()
```

# **Data Analysis**

```python
# Step 12: Check Fuel Types
df["Fuel_Type"].value_counts()
```

```python
# Step 13: Fuel Type Count Graph
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Fuel_Type")
plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.show()
```

```python
# Step 14: Selling Price Distribution
plt.figure(figsize=(8,5))
plt.hist(df["Selling_Price"], bins=20)
plt.title("Selling Price Distribution")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")
plt.show()
```

```python
# Step 15: Present Price vs Selling Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Present_Price", y="Selling_Price")
plt.title("Present Price vs Selling Price")
plt.xlabel("Present Price")
plt.ylabel("Selling Price")
plt.show()
```

```python
# Step 16: Kms Driven vs Selling Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Kms_Driven", y="Selling_Price")
plt.title("Kilometers Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price")
plt.show()
```

```python
# Step 17: Selling Price by Fuel Type
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Fuel_Type", y="Selling_Price")
plt.title("Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Selling Price")
plt.show()
```

```python
# Step 18: Selling Price by Transmission
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Transmission", y="Selling_Price")
plt.title("Selling Price by Transmission Type")
plt.xlabel("Transmission")
plt.ylabel("Selling Price")
plt.show()
```

```python
# Step 19: Selling Price by Seller Type
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Seller_Type", y="Selling_Price")
plt.title("Selling Price by Seller Type")
plt.xlabel("Seller Type")
plt.ylabel("Selling Price")
plt.show()
```

```python
# Step 20: Year vs Selling Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Year", y="Selling_Price")
plt.title("Manufacturing Year vs Selling Price")
plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price")
plt.show()
```

```python
# Step 21: Correlation Analysis
numeric_df = df.select_dtypes(include=np.number)
correlation = numeric_df.corr()
display(correlation)
```

```python
# Step 22: Correlation Heatmap
plt.figure(figsize=(10,7))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Step 23: Top 10 Car Models
df["Car_Name"].value_counts().head(10)
```

```python
# Step 24: Top 10 Car Models Graph
top_cars = df["Car_Name"].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_cars.values, y=top_cars.index)
plt.title("Top 10 Most Common Car Models")
plt.xlabel("Number of Cars")
plt.ylabel("Car Name")
plt.show()
```

```python
# Step 25: Average Selling Price by Fuel Type
fuel_price = df.groupby("Fuel_Type")["Selling_Price"].mean()
display(fuel_price)
```

```python
# Step 26: Average Selling Price Graph

plt.figure(figsize=(8,5))
fuel_price.plot(kind="bar")
plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=0)
plt.show()
```

```python
# Step 27: Average Selling Price by Transmission

transmission_price = df.groupby("Transmission")["Selling_Price"].mean()
display(transmission_price)
```

```python
# Step 28: Average Selling Price by Seller Type
seller_price = df.groupby("Seller_Type")["Selling_Price"].mean()
display(seller_price)
```

```python
# Step 29: Find Most Expensive Car
df.loc[df["Selling_Price"].idxmax()]
```

```python
# Step 30: Find Cheapest Car
df.loc[df["Selling_Price"].idxmin()]
```

```python
# Step 31: Basic Findings
print("Total Cars:", len(df))
print("\nAverage Selling Price:", round(df["Selling_Price"].mean(), 2))
print("\nAverage Present Price:", round(df["Present_Price"].mean(), 2))
print("\nAverage Kilometers Driven:", round(df["Kms_Driven"].mean(), 2))
```

### CONCLUSION

1. Car prices vary according to present price, age, kilometers driven, fuel type, transmission and seller type.
2. Present Price and Selling Price show an important relationship.
3. Older cars generally have lower selling prices.
4. Cars with different fuel types and transmission types show different price patterns.
5. Data analysis helps buyers and sellers understand used-car market trends.
6. The analysis can support better car buying, selling and pricing decisions.
